```python
import chromadb
print(chromadb.__version__)

client = chromadb.Client()
#to save on disk use .PersistentClient(path = "./chroma_data")
#there are also Http Client and Cloud Client look at the last section of the intro video.

collection = client.get_or_create_collection(name="my_collection")

collection.update( #add

     ids = ["id1", "id2"] ,
     documents = ["this is a document about pineapple",
                  "this is a document about orange"],
     metadatas=[{"category": "fruit doc"}, {"category": "fruit doc"}]
)

results = collection.query (query_texts=["this is a query document about florida"], n_results=2)
print(results)

#print(collection.peek())
```
